## Example - Linear Programming and MOOP

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.lines import Line2D
```

```python
# Set a consistent Seaborn style

sns.set_theme(style="whitegrid")

# Generate data for the feasible region
d = np.linspace(-2, 5, 1000)  # Extended range to ensure all points are visible
x, y = np.meshgrid(d, d)
feasible_region = ((y >= 0) & (x >= 0) & (x + y >= 1) & (x - y <= 2) & (x + 2 * y <= 5)).astype(int)

# Plot the feasible region using a filled contour plot
plt.figure(figsize=(12, 8))  # Widened the plot
plt.contourf(x, y, feasible_region, levels=[-0.5, 0.5, 1.5], colors=["white", "orange"], alpha=0.3)

# Generate data for the constraint lines
x_line = np.linspace(-1, 5, 2000)
y3 = 1 - x_line  # x1 + x2 >= 1
y4 = x_line - 2  # x1 - x2 <= 2
y5 = (5 - x_line) / 2  # x1 + 2*x2 <= 5

# Plot the constraint lines with the same color (black)
sns.lineplot(x=x_line, y=y3, color="black", linestyle="-")
sns.lineplot(x=x_line, y=y4, color="black", linestyle="-")
sns.lineplot(x=x_line, y=y5, color="black", linestyle="-")

# Add x >= 0 and y >= 0 lines with increased thickness
sns.lineplot(x=[0, 0], y=[-0.5, 3.5], color="black", linestyle="-", linewidth=2, label=r"$x_1 \geq 0$")
sns.lineplot(x=[-0.5, 5], y=[0, 0], color="black", linestyle="-", linewidth=2, label=r"$x_2 \geq 0$")

# Plot the points with larger markers
points = [(0, 1), (1, 0), (2, 0), (0, 2.5), (3, 1)]
colors = ['green', 'blue', 'blue', 'blue', 'blue']
markers = ['o', 'X', 'X', 'X', 'X']

for point, color, marker in zip(points, colors, markers):
    plt.scatter(point[0], point[1], color=color, s=200, marker=marker)  # Increased marker size to 200

# Set plot limits to ensure all points and lines are visible
plt.xlim(-0.5, 5)
plt.ylim(-0.5, 3.5)

# Explicitly set ticks to include 0
plt.xticks(np.arange(0, 6, 1),fontsize=16)
plt.yticks(np.arange(0, 4, 1),fontsize=16)

# Set axis labels and title
plt.xlabel(r'$x_1$', fontsize=18)
plt.ylabel(r'$x_2$', fontsize=18)
plt.title("Feasible Region, Constraints, and Points", fontsize=20, pad=20)

# Replace legend with a custom one
custom_legend = [
    Line2D([0], [0], marker='o', color='w', markerfacecolor='green', markersize=10, label='Optimal'),
    Line2D([0], [0], marker='X', color='w', markerfacecolor='blue', markersize=10, label='Not Optimal')
]
plt.legend(handles=custom_legend, loc='upper right', fontsize=14, frameon=True, shadow=True)

# Add gridlines for better readability
plt.grid(color='gray', linestyle='--', linewidth=0.5, alpha=0.7)

# Ensure grid regions are square
plt.gca().set_aspect('equal', adjustable='box')

# Show the plot
plt.tight_layout()
plt.show()
```

```python
from KM_Algo import KungMethod as KM
from Functions import Function
from Point import Point

# Generate an array of points within the feasible region
def f1(x):
    return 3*x[0] + 2*x[1]

def f2(x):
    return 7 - x[0] - x[1]

func_1 = Function(f1,f1,"f1")
func_2 = Function(f2,f2,"f2")
obj_functions = [func_1, func_2]

```

```python
points = []
for i in range(-500,3000):
    for j in range(-500,3000):
        x = i/3000 * 5
        y = j/3000 * 3.5
        if ((y >= 0) & (x >= 0) & (x + y >= 1) & (x - y <= 2) & (x + 2 * y <= 5)):
            points.append([x,y])

POINTSLIST = []
for p in points:
    newPoint = Point(p)
    newPoint.evaluate(obj_functions)
    POINTSLIST.append(newPoint)

Results = KM(POINTSLIST.copy())

# Extract vectors from results
vectors_x = [point.vector[0] for point in Results]
vectors_y = [point.vector[1] for point in Results]

pointslist_x = [point.eval_f[0] for point in POINTSLIST]
pointslist_y = [point.eval_f[1] for point in POINTSLIST]
```

```python
# Extract coordinates for Results and POINTSLIST
results_x = [point.eval_f[0] for point in Results]
results_y = [point.eval_f[1] for point in Results]

# Create the plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x=pointslist_x, y=pointslist_y, color='blue', s=10, label='POINTSLIST')
sns.scatterplot(x=results_x, y=results_y, color='red', s=50, label='Results')

# Add labels, legend, and title
plt.xlabel('Objective 1', fontsize=14)
plt.ylabel('Objective 2', fontsize=14)
plt.title('Results vs POINTSLIST', fontsize=16)
plt.legend(fontsize=12)
plt.grid(alpha=0.3)

# Show the plot
plt.tight_layout()
plt.show()


```
