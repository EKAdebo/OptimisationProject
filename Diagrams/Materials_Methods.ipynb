```python
def f1(x):
    return x[0]**2 + x[1]**2

func_1 = Function(f1,f1,"f1")

def f2(x):
    return (x[0] -3)**2 + (x[1] - 4)**2


func_2 = Function(f2,f2,"f2")

obj_functions = [func_1, func_2]
```

```python
from KM_Algo import KungMethod


n = 1000
d = np.linspace(-0.5, 5, n)
x1, x2= np.meshgrid(d, d)

valid = []


for i in range(n):
    for j in range(n):
        if (d[i] + d[j] <= 5)&(d[i] >= 0)&(d[j] >= 0) &(d[i] - d[j] >= 1):
            x = Point((d[i],d[j]))
            x.evaluate(obj_functions)
            valid.append(x)

km = valid.copy()

km = KungMethod(km)

```

```python
from scipy.interpolate import PchipInterpolator

# Remove duplicates from x_points and y_points

# Create an interpolation function
# Example data points
x_points = [x.vector[0] for x in km]
y_points = [x.vector[1] for x in km]


import numpy as np
from scipy.interpolate import interp1d
import matplotlib.pyplot as plt
from collections import defaultdict

def average_y_for_duplicate_x(x_list, y_list):
    """
    Handle duplicate x-values by averaging the corresponding y-values.

    Args:
        x_list (list): List of x-coordinates.
        y_list (list): List of y-coordinates.

    Returns:
        Tuple of two lists: Processed x-coordinates and corresponding averaged y-values.
    """
    # Use a defaultdict to group y-values by x-values
    point_dict = defaultdict(list)
    for x, y in zip(x_list, y_list):
        point_dict[x].append(y)

    # Average the y-values for each unique x-value
    processed_x = []
    processed_y = []
    for x, y_values in point_dict.items():
        processed_x.append(x)
        processed_y.append(np.mean(y_values))

    # Return the sorted x and y lists
    sorted_indices = np.argsort(processed_x)
    processed_x = np.array(processed_x)[sorted_indices].tolist()
    processed_y = np.array(processed_y)[sorted_indices].tolist()

    return processed_x, processed_y


x_og = [x.vector[0] for x in valid]
y_og = [x.vector[1] for x in valid]

#plt.scatter(x_og,y_og, color='red')

processed_x, processed_y = average_y_for_duplicate_x(x_points, y_points)

# Create an interpolation function
interp_func = interp1d(processed_x, processed_y,kind = "linear",fill_value='extrapolate')

# Generate new x values for the curve
x_new = np.linspace(1,3, 1000)
y_new = interp_func(x_new)

plot_feasible(d,x1,x2,False)

# Plot the original points and the interpolated curve
#plt.plot(x_points, y_points, 'o', label='Original points')
plt.plot(x_new, y_new, '-',color ='red', label='Pareto Front')




k1 = [x.eval_f[0] for x in km]
k2 = [x.eval_f[1] for x in km]



plt.legend(loc='upper right')

plt.show()
```
